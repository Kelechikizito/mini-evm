```python
# All the EVM opcodes are defined here. All opcodes have one thing in common. They manipulate the current state of the EVM.
# STOP
STOP = 0x0 # This halts execution by setting the stop_flag to True.

# MATH: These are opcodes for mathematical operations in the EVM.
ADD = 0x1 # addition
MUL = 0x2 # multiplication
SUB = 0x3 # subtraction
DIV = 0x4 # division
SDIV = 0x5 # signed integer division
MOD = 0x6 # modulo operator
SMOD = 0x7 # signed modulo operator
ADDMOD = 0x8 # modulo addition
MULMOD = 0x9 # modulo multiplication
EXP = 0xA # Exponential operation
SIGNEXTEND = 0xB # Extend length of two’s complement signed integer

# COMPARISONS: These opcodes are used for comparisons
LT = 0x10 # Less than <
GT = 0x11 # Greater than >
SLT = 0x12 # Signed less than
SGT = 0x13 # Signed greater than
EQ = 0x14 # Equality sign ==
ISZERO = 0x15 # Is-zero comparison

# LOGIC: These opcodes are used for logical operations
AND = 0x16 # &&
OR = 0x17 # |
XOR = 0x18 # Exclusive OR
NOT = 0x19 # NOT

# BIT OPERATIONS: These opcodes are used for bit operations
BYTE = 0x1A # Retrieve a single byte from a word
SHL = 0x1B # Shift-left operation
SHR = 0x1C # Shift-right operation
SAR = 0x1D # Arithmetic (signed) right shift operation

# MISCELLANEOUS
SHA3 = 0x20 # Compute Keccak-256 hash

# ETHEREUM STATE: These opcodes give you access to the Ethereum environment
ADDRESS = 0x30 # Get address of currently executing account
BALANCE = 0x31 # Get balance of a given account
ORIGIN = 0x32 # Get execution origination address
CALLER = 0x33 # Get caller address
CALLVALUE = 0x34 # Get deposited value by the instruction/transaction responsible for this execution
CALLDATALOAD = 0x35 # Get the data payload of a call
CALLDATASIZE = 0x36 # Get the size of the data payload of a call
CALLDATACOPY = 0x37 # Copy input data in current environment to memory
CODESIZE = 0x38 # Get the size of code in the current working environment
CODECOPY = 0x39 # Copy code in current environment to memory
GASPRICE = 0x3A # Get the price of gas in current environment
EXTCODESIZE = 0x3B # Get size of an account’s code
EXTCODECOPY = 0x3C # Copy an account’s code to memory
RETURNDATASIZE = 0x3D # Get size of output data from the previous call from the current environment
RETURNDATACOPY = 0x3E # Copy output data from the previous call to memory
EXTCODEHASH = 0x3F # Get hash of an account's code
BLOCKHASH = 0x40 # Get the hash of one of the 256 most recent complete blocks
COINBASE = 0x41 # Get the block’s beneficiary/VALIDATOR/MINER address
TIMESTAMP = 0x42 # Get the timestamp of a block
NUMBER = 0x43 # Get the block's number
PREVRANDAO = 0x44 # Get the previous block’s RANDAO mix
GASLIMIT = 0x45 # Get the block's gas limit
CHAINID = 0x46 # Get the current chainId
SELFBALANCE = 0x47 # Get balance of currently executing account
```
