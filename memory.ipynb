```python
class SimpleMemory:
    def __init__(self):
        self.memory = [] # This line initiates an empty memory list

    def access(self, offset, size):  # This line defines how to access the memory list; offset being the index WHILE the size being the size. An offset of 2 would mean that we get the byte stored at index 2. Combined with a size we can get a block of bytes. Offset 2 and size 5 would return the bytes from index 2 to index 6.
        return self.memory[offset : offset + size] # This line returns a slice of the memory list based on the given offset and size    

    def load(self, offset): # This line defines how to load 32 bytes
        return self.access(offset, 32) # This line returns the result of the access method with a size of 32 bytes

    def store(self, offset, value):
        self.memory[offset: offset + len(value)] = value
```

```python
class Memory(SimpleMemory):
    def store(self, offset, value):
        memory_expansion_cost = 0  # This line initializes the variable to track gas cost for memory expansion

        if len(self.memory) <= offset + len(
            value
        ):  # This line checks if the current memory size is less than or equal to the required size after storing the new value
            expansion_size = 0  # This line initializes the variable to calculate how much memory needs to be expanded

            if (
                len(self.memory) == 0
            ):  # If the memory is empty, expand the memory size to 32 bytes
                expansion_size = 32
                self.memory = [
                    0x00 for _ in range(expansion_size)
                ]  # This line expands the memory to 32 bytes by filling it with zeros

            if len(self.memory) < offset + len(
                value
            ):  # This line checks again if the memory size is still insufficient after the initial expansion
                expansion_size += (
                    offset + len(value) - len(self.memory)
                )  # This line calculates the additional expansion size needed to accommodate the new value
                self.memory.extend(
                    [0x00] * expansion_size
                )  # This line extends the memory by adding the required number of zero bytes

            memory_expansion_cost = (
                expansion_size**2
            )  # This line calculates the gas cost for memory expansion based on the size of the expansion

        super().store(offset, value)  # This line calls the store method of the parent class to actually store the value in memory
        return memory_expansion_cost  # This line returns the calculated gas cost for memory expansion
```
