```python
# The EVM stack has a maximum capacity of 1024 items. Every item on the stack is at max a 256-bit value (32 bytes)
MAXIMUM_STACK_SIZE = 1024

class Stack:
    # The __init__ dunder is the constructor method to initialize new objects of the class, in this case, empty items list.
    def __init__(self):
        self.items = [] # The items list holds all the stack items

    def __str__(self,): # This dunder method converts the output of the stack to a string
        ws = [] # This is a temporary list to hold the string representations of the stack items
        for i, item in enumerate(self.items[::-1]): # This loops through the stack items in reverse order, where i is the index and item is the value of the stack item
            if i == 0: # if the index is 0, it means it's the top of the stack
                ws.append(f"{item} <first") 
            elif i == len(self.items) - 1: # if the index is the last one, it means it's the bottom of the stack
                ws.append(f"{item} <last")
            else: # for all other items in the stack, we just append their string representation
                ws.append(str(item))
        return "\n".join(ws) # Finally, we join all the string representations with a newline character and return the resulting string

    def push(self, value):
        if len(self.items) == MAXIMUM_STACK_SIZE - 1: # This checks if the stack items is at its maximum capacity, if it is, it raises an exception
            raise Exception("Stack overflow")
        self.items.append(value) # If the stack is not full, it appends the new

    def pop(self):
        if len(self.items) == 0:
            raise Exception("Stack underflow") # This checks if the stack items are zero so it won't pop nothing, thereby wasting gas.
        return self.items.pop()

    @property
    def stack(self):
        return self.items.copy() # This is a property method that returns a copy of the stack items list, allowing external code to access the stack without modifying the original list.
```
